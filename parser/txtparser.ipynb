```python

# load txt file
with open("text_reference.txt", "r") as f:
    text = f.read()

# parse txt file
# load each line and extract the reference
# save the reference to a new file
# with open("text_reference.txt", "r") as f:
#     for line in f:
#         reference = line.split("]")[1]
#         print(reference)
# f.close()
cnt = 0
# save the reference in a dict
reference_dict = {}
with open("text_reference.txt", "r") as f:
    for line in f:
        if "arXiv:" in line:
            reference = line.split("arXiv:")[1].split("*")[0]
            cnt += 1
            reference_dict[cnt] = reference
f.close()

print(reference_dict)
```

```python
import subprocess

# use wget to download the reference
for key, value in reference_dict.items():
    url = "https://arxiv.org/pdf/" + value
    subprocess.run(["wget", "-P", "/root/autodl-tmp/Code/autosurvey/AutoSurvey/pdf", url])

```
