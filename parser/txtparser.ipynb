```python
import requests
import xml.etree.ElementTree as ET

# arXiv API 请求 URL
paper_id = "2301.12345"  # 替换为你的论文编号
url = f"http://export.arxiv.org/api/query?id_list={paper_id}"

# 发送请求
response = requests.get(url)

# 检查请求是否成功
if response.status_code == 200:
    # 解析 XML 数据
    root = ET.fromstring(response.content)

    # 提取标题和摘要
    for entry in root.findall("{http://www.w3.org/2005/Atom}entry"):
        title = entry.find("{http://www.w3.org/2005/Atom}title").text
        abstract = entry.find("{http://www.w3.org/2005/Atom}summary").text
        pdf_link = entry.find("{http://www.w3.org/2005/Atom}link[@title='pdf']").attrib['href']

        print(f"Title: {title}")
        print(f"Abstract: {abstract}")
        print(f"PDF Link: {pdf_link}")
else:
    print(f"Failed to fetch data. Status code: {response.status_code}")
```

```python
import os
import re
import requests
import json

# Directory containing the markdown files
directory = "/root/autodl-tmp/Code/autosurvey/AutoSurvey/docs/md"

# Regular expression to match filenames like xxxx.xxxxx.md
pattern = re.compile(r"(\d{4}\.\d{5})\.md")

# arXiv API endpoint
arxiv_api_url = "http://export.arxiv.org/api/query"

# Dictionary to store paper details
papers = {}

# Iterate over all files in the directory
for filename in os.listdir(directory):
    match = pattern.match(filename)
    if match:
        paper_id = match.group(1)
        file_text = ""
        with open(directory + "/" + filename, "r") as f:
            file_text = f.read()

        # get the text after "introduction"
        index = file_text.find("Introduction")
        if index == -1:
            index = file_text.find("INTRODUCTION")
        if index == -1:
            index = file_text.find("introduction")

        if index != -1:
            file_text = file_text[index:index+1500]
        else:
            file_text = ""

        # Query arXiv API
        params = {
            "id_list": paper_id,
            "max_results": 1
        }
        response = requests.get(arxiv_api_url, params=params)
        if response.status_code == 200:
            # Parse the response XML
            from xml.etree import ElementTree as ET
            root = ET.fromstring(response.content)
            
            # Extract title and abstract
            entry = root.find("{http://www.w3.org/2005/Atom}entry")
            if entry is not None:
                title = entry.find("{http://www.w3.org/2005/Atom}title").text.strip()
                abstract = entry.find("{http://www.w3.org/2005/Atom}summary").text.strip()
                
                # Save to dictionary
                papers[paper_id] = {
                    "title": title,
                    "abstract": abstract,
                    "introduction": file_text
                }
        else:
            print(f"Failed to fetch data for paper {paper_id}")

# Save the results to a JSON file
output_file = "papers.json"
with open(output_file, "w") as f:
    json.dump(papers, f, indent=4)

print(f"Saved {len(papers)} papers to {output_file}")
```

```python
from transformers import AutoTokenizer, AutoModel
import torch
import numpy as np
import faiss

# 加载模型和分词器
model_name = "nomic-ai/nomic-embed-text-v1"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

# 示例数据
articles = [
    {"title": "Deep Learning for Natural Language Processing", "abstract": "A comprehensive survey of deep learning techniques in NLP."},
    {"title": "Attention Mechanisms in Neural Networks", "abstract": "Exploring the role of attention in modern neural architectures."},
    # 添加更多文章
]

# 生成嵌入向量
def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True, max_length=512)
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state[:, 0, :].squeeze().numpy()

# 为每篇文章生成嵌入
embeddings = []
for article in articles:
    text = f"{article['title']} {article['abstract']}"
    embedding = get_embedding(text)
    embeddings.append(embedding)
embeddings = np.array(embeddings)

# 创建 Faiss 索引
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

# 示例查询
query_text = "How does attention work in neural networks?"
query_embedding = get_embedding(query_text).reshape(1, -1)

# 搜索最相似的 5 篇文章
k = 5
distances, indices = index.search(query_embedding, k)

# 输出结果
for i, idx in enumerate(indices[0]):
    print(f"Result {i + 1}:")
    print(f"Title: {articles[idx]['title']}")
    print(f"Abstract: {articles[idx]['abstract']}")
    print(f"Distance: {distances[0][i]}")
    print("-" * 50)
```
